```python
dataset.loc[:122,'Region'] = 0
dataset.loc[122:,'Region'] = 1
df=dataset
df.info()
```

```python
df[['Region']]=df[['Region']].astype('int64')
df.info()
```

```python
df.isnull().sum()
```

```python
df=df.dropna().reset_index(drop=True)
df.isnull().sum()
```

```python
df.iloc[[122]]
```

```python
df=df.drop(122).reset_index(drop=True)
df.iloc[[122]]
```

```python
df.columns
```

```python
df.columns=df.columns.str.strip()
df.columns
```

```python
df.info()
```

```python
df.head()
```

```python
df[['month','day','year','Temperature','RH', 'Ws']]=df[['month','day','year','Temperature','RH', 'Ws']].astype('int')
df.info()
```

```python
objects=[features for features in df.columns if df[features].dtype=='string']
```

```python
for i in objects:
    if i!='Classes':
        df[i]=df[i].astype(float)
    else:
        df[i]=df[i].astype(object)        
```

```python
plt.style.use('seaborn-v0_8')
df_copy.hist(bins=50, figsize=(20,15))
plt.show()
```

```python
percentage=df_copy['Classes'].value_counts(normalize=True)* 100
```

```python
classlabels=['Not Fire','Fire']
plt.figure(figsize=(12,7))
plt.pie(percentage, labels=classlabels, autopct='%1.1f%%', startangle=90)
plt.title('Distribution of Classes')
plt.show()
```

```python
df['Classes']=np.where(df['Classes'].str.contains('not fire'),'not fire','fire')
```

```python
dftemp=df.loc[df['Region']==1]
plt.subplots(figsize=(13,6))
sns.set_style('whitegrid')
sns.countplot(x='month', hue='Classes', data=dftemp, palette='Set2')
plt.ylabel('Number of Fires', weight='bold')
plt.xlabel('Month', weight='bold')
plt.title('Distribution of Classes by Month for Region 1')
```
