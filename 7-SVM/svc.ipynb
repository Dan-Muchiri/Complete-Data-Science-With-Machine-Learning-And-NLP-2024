```python
import pandas as pd
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
from sklearn.datasets import make_classification
```

```python
x,y=make_classification(n_samples=1000,n_features=2, n_classes=2, n_clusters_per_class=2, n_redundant=0)
```

```python
pd.DataFrame(x)[0]
```

```python
sns.scatterplot(x=pd.DataFrame(x)[0],y=pd.DataFrame(x)[1],hue=y)
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.25,random_state=42)
```

```python
from sklearn.svm import SVC
svc=SVC(kernel='linear')
svc.fit(x_train,y_train)
```

```python
y_pred=svc.predict(x_test)
```

```python
from sklearn.metrics import classification_report,confusion_matrix
print(classification_report(y_test,y_pred))
print(confusion_matrix(y_test,y_pred))
```

```python
rbf=SVC(kernel='rbf')
rbf.fit(x_train,y_train)
```

```python
rbf_pred=rbf.predict(x_test)
print(classification_report(y_test,rbf_pred))
print(confusion_matrix(y_test,rbf_pred))
```

```python
polynomial=SVC(kernel='poly')
polynomial.fit(x_train,y_train)
polynomial_pred=polynomial.predict(x_test)
print(classification_report(y_test,polynomial_pred))
print(confusion_matrix(y_test,polynomial_pred))
```

```python
sigmodial=SVC(kernel='sigmoid')
sigmodial.fit(x_train,y_train)
sigmodial_pred=sigmodial.predict(x_test)
print(classification_report(y_test,sigmodial_pred))
print(confusion_matrix(y_test,sigmodial_pred))
```

```python
from sklearn.model_selection import GridSearchCV
param_grid={'C':[0.1,1,10,100],'gamma':[1,0.1,0.01,0.001],'kernel':['rbf']}
grid=GridSearchCV(SVC(),param_grid,refit=True,verbose=3, cv=5)
grid.fit(x_train,y_train)
```

```python
grid.best_params_
```

```python
grid_pred=grid.predict(x_test)
print(classification_report(y_test,grid_pred))
print(confusion_matrix(y_test,grid_pred))
```
